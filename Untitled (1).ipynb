```python
feedbacks = [
    "The service was great",
    "Great experience overall",
    "Bad product, terrible service",
    "It was okay, not great"
]

def count_word_in_feedbacks(word):
    target = word.lower()
    return sum(1 for fb in feedbacks if target in fb.lower())
```

```python
def print_feedback_counts(feedback_data):
    # Count how many strings contain the target words
    good_count = sum(1 for f in feedback_data if "good" in f.lower())
    poor_count = sum(1 for f in feedback_data if "poor" in f.lower())
    excellent_count = sum(1 for f in feedback_data if "excellent" in f.lower())

    # Print the requested counts
    print(f"● Number of feedback containing \"good\": {good_count}")
    print(f"● Number of feedback containing \"poor\": {poor_count}")
    print(f"● Number of feedbacks containing \"excellent\": {excellent_count}")

# Example usage:
feedback_list = [
    "The service was good and fast.",
    "Poor experience, very slow.",
    "This is an excellent product!",
    "Not bad, pretty good overall.",
    "An excellent choice, highly recommended."
]

print_feedback_counts(feedback_list)
```

```python
import re
feedback_data = {
    "rating": [5, 3, 4, 2, 5],
    "comment": [
        "Excellent product and fast shipping!",
        "It was okay but arrived late.",
        "Good quality for the price.",
        "Very disappointed with the customer service.",
        "Absolutely amazing experience overall."
    ]
}

# 1. Display the final cleaned feedback_data
print("--- Final Cleaned Feedback Data ---")
print(feedback_data)

# 2. Print the average rating
avg_rating = sum(feedback_data["rating"]) / len(feedback_data["rating"])
print(f"\n--- Average Rating ---\n{avg_rating:.2f}")

# 3. Find and display the feedback with the longest comment (by word count)
longest_comment = max(feedback_data["comment"], key=lambda c: len(c.split()))
print(f"\n--- Longest Comment ---\n\"{longest_comment}\"")

# 4. Print the list of unique words used across all feedbacks
all_comments_text = " ".join(feedback_data["comment"]).lower()
# re.findall extracts only words, ignoring punctuation marks like ! and .
all_words = re.findall(r'\b\w+\b', all_comments_text)
unique_words = sorted(list(set(all_words)))
print(f"\n--- Unique Words ---\n{unique_words}")

# 5. Optional: Sort feedbacks by rating (highest to lowest) using zip() and sorted()
zipped_feedback = list(zip(feedback_data["rating"], feedback_data["comment"]))
sorted_feedback = sorted(zipped_feedback, key=lambda x: x[0], reverse=True)

print("\n--- Sorted Feedbacks (Highest to Lowest) ---")
for rating, comment in sorted_feedback:
    print(f"Rating: {rating} | Comment: {comment}")
```
